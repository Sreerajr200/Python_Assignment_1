3. String Methods:
Write a Python program that takes a string, strM = “Python beginner tutorial” and
perform the following tasks:
a. Convert the sentence to uppercase.
b. Convert the sentence to lowercase.
c. Use Capitalize and return the sentence to the original input form.
d. Count the total number of occurrences of character ‘t’ in the string.
e. Replace all occurrences of “Python” with “Machine Learning” in the input string
strM = “Python beginner tutorial”
************************************************************************************************************************************

```python
strM = "Python beginner tutorial"
print("Convert the sentence to uppercase - ",strM.upper())
print("Convert the sentence to lowercase - ",strM.lower())
print("Use Capitalize and return the sentence to the original input form - ",strM.capitalize())
print("Count the total number of occurrences of character ‘t’ in the string - ",strM.count('t'))
print("Replace all occurrences of “Python” with “Machine Learning” in the input string - ",strM.replace('Python','Machine Learning'))
```

Tuples (Creation, Modification and Access) :
Create 1st tuple with values -> (10, 20, 30), 2nd tuple with values -> (40, 50, 60):
a. Concatenate the two tuples and store it in “t_combine”
b. Repeat the elements of “t_combine” 3 times
c. Access the 3rd element from “t_combine”
d. Access the first three elements from “t_combine”
e. Access the last three elements from “t_combine”
************************************************************************************************************************************

```python
Tup1 = (10,20,30)
Tup2 = (40,50,60)
t_combine = Tup1+Tup2
print("Concatenate the two tuples and store it in t_combine - ", t_combine)
print("Repeat the elements of t_combine 3 times - ", t_combine*3 )
print("Access the 3rd element from t_combine - ", t_combine[2])
print("Access the first three elements from t_combine - ",t_combine[:3] )
print("Access the last three elements from t_combine - ",t_combine[-3:] )
```
